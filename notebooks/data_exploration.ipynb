```python
import pandas as pd
```

```python
datasets = pd.read_csv('../data/raw/loan_data.csv')
datasets.head(10)
```

```python
print(datasets["Gender"].value_counts())
print(datasets["Education"].value_counts())
print(datasets["Dependents"].value_counts())
print(datasets["Self_Employed"].value_counts())
print(datasets["Married"].value_counts())
```

```python
from sklearn.impute import SimpleImputer
imputer = SimpleImputer(strategy='most_frequent')
```

```python
datasets["Gender"] = imputer.fit_transform(datasets[["Gender"]]).ravel()
datasets['Married'] = imputer.fit_transform(datasets[['Married']]).ravel()
datasets['Dependents'] = imputer.fit_transform(datasets[['Dependents']]).ravel()
datasets['Self_Employed'] = imputer.fit_transform(datasets[['Self_Employed']]).ravel()
datasets['LoanAmount'] = imputer.fit_transform(datasets[['LoanAmount']]).ravel()
datasets['Loan_Amount_Term'] = imputer.fit_transform(datasets[['Loan_Amount_Term']]).ravel()
```

```python
datasets.isnull().sum()
```

```python
datasets.head(30)
```

```python
print(datasets["Gender"].value_counts())
print(datasets["Education"].value_counts())
print(datasets["Dependents"].value_counts())
print(datasets["Self_Employed"].value_counts())
print(datasets["Married"].value_counts())
print(datasets["Property_Area"].value_counts())
print(datasets["Loan_Status"].value_counts())
print(datasets["Credit_History"].value_counts())
```

```python
from sklearn.preprocessing import LabelEncoder, OneHotEncoder, StandardScaler

labelencoder = LabelEncoder()
hotencoder = OneHotEncoder()
scaler = StandardScaler()
```

```python
datasets["Gender"] = labelencoder.fit_transform(datasets["Gender"])
mapping = {label: idx for idx, label in enumerate(labelencoder.classes_)}
print(mapping, "Gender mapping \n")

datasets["Married"] = labelencoder.fit_transform(datasets["Married"])
mapping = {label: idx for idx, label in enumerate(labelencoder.classes_)}
print(mapping, "Married mapping \n")

datasets["Education"] = labelencoder.fit_transform(datasets["Education"])
mapping = {label: idx for idx, label in enumerate(labelencoder.classes_)}
print(mapping, "Education mapping \n")

datasets["Self_Employed"] = labelencoder.fit_transform(datasets["Self_Employed"])
mapping = {label: idx for idx, label in enumerate(labelencoder.classes_)}
print(mapping, "Self_Employed mapping \n")
```

```python
datasets.head(20)
```

```python
# encode property area too
datasets["Property_Area"] = labelencoder.fit_transform(datasets["Property_Area"])
mapping = {label: idx for idx, label in enumerate(labelencoder.classes_)}
print(mapping, "Property_Area mapping \n")
```

```python
datasets.head(20)
```

```python
# hot encode dependents
Dependents_onehot = hotencoder.fit_transform(datasets[["Dependents"]]).toarray()
```

```python
Dependents_datasets = pd.DataFrame(Dependents_onehot, columns=hotencoder.get_feature_names_out(["Dependents"]))
print(Dependents_datasets, "Dependents encoded \n")
```

```python
Dependents_datasets.head()
```

```python
datasets = pd.concat([datasets, Dependents_datasets], axis=1)
```

```python
datasets.head()
```

```python
# encode loan status
datasets["Loan_Status"] = labelencoder.fit_transform(datasets["Loan_Status"])
```

```python
datasets.head()
```

```python
# get the mappings for all the encoded columns (for gender, it should be {"Male": 1, "Female": 0})
print(datasets["Gender"].value_counts())
```

```python
# drop loanid and dependents
datasets = datasets.drop(["Loan_ID", "Dependents", "CoapplicantIncome", "Credit_History"], axis=1)
datasets.head()
```

```python
# scale the loan amount and loan amount term
datasets[["LoanAmount"]] = scaler.fit_transform(datasets[["LoanAmount"]])
datasets[["Loan_Amount_Term"]] = scaler.fit_transform(datasets[["Loan_Amount_Term"]])
datasets[["ApplicantIncome"]] = scaler.fit_transform(datasets[["ApplicantIncome"]])
```

```python
datasets.head()
```

```python
# save to csv
datasets.to_csv('../data/processed/loan_data_processed.csv', index=False)
```
